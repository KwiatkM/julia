```julia
using JLD2
X_train = load("C:\\Users\\kwiat\\projekty\\julia\\my_AD_library\\data\\imdb_dataset_prepared.jld2", "X_train")
y_train = load("C:\\Users\\kwiat\\projekty\\julia\\my_AD_library\\data\\imdb_dataset_prepared.jld2", "y_train")
X_test = load("C:\\Users\\kwiat\\projekty\\julia\\my_AD_library\\data\\imdb_dataset_prepared.jld2", "X_test")
y_test = load("C:\\Users\\kwiat\\projekty\\julia\\my_AD_library\\data\\imdb_dataset_prepared.jld2", "y_test")
nothing
```

```julia
X_train = permutedims(X_train', (2,1))
size(X_train)
```

```julia
typeof(X_train)
```

```julia
y_train = permutedims(y_train,(2,1))
size(y_train)
```

```julia
typeof(y_train)
```

```julia
y_train = Float32.(y_train)
typeof(y_train)
```

```julia
model = buildModel(Float64,
    FullyConnected(size(X_train, 1),32,ReLU),
    FullyConnected(32,1,sigm),
    Loss(myBinaryCrossEntropy)
    )
nothing
```

```julia
options = Options();
options.nEpochs = 10;
options.batchSize = 64;
setDefaultAdamOptions(options)

optimizer = Adam(options,model);
```

```julia
train(model, X_train, y_train, options, optimizer)
```

```julia
println("Real value: ", y_train[1])
println("Inference value: ", inference(model,X_train[:,1]))
println("Real value: ", y_train[2])
println("Inference value: ", inference(model,X_train[:,2]))
println("Real value: ", y_train[3])
println("Inference value: ", inference(model,X_train[:,3]))
println("Real value: ", y_train[4])
println("Inference value: ", inference(model,X_train[:,4]))
println("Real value: ", y_train[5])
println("Inference value: ", inference(model,X_train[:,5]))
```

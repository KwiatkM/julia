```julia
y_train = Float32.(y_train)
y_train = permutedims(y_train,(2,1))
size(y_train)
```

```julia
model = buildModel(Float32, Int,
    Embedding(130, embeddings),
    PermuteDims((2,1,3)),
    Convolution_1d(embedding_dim, 8, 3, ReLU),
    MaxPool_1d(8),
    myFlatten(),
    FullyConnected(128,1,sigm),
    Loss(myBinaryCrossEntropy)
    )
nothing
```

```julia
options = Options();
options.nEpochs = 15;
options.batchSize = 64;
setDefaultAdamOptions(options)
optimizer = Adam(options,model);
```

```julia
train(model, X_train, y_train, options, optimizer)
```
